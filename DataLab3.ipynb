## Step 4
Now that our libraries are imported, let's put yfinance to work. We are going to download the daily stock prices for Microsoft (ticker symbol **MSFT**) and Alphabet, the parent company of Google (ticker symbol **GOOG**), from January 1, 2020 through December 31, 2025. We'll save the Microsoft data as **dfM** and the Google data as **dfG**. The "df" stands for **data frame**, which is what pandas calls a table of data.  We could call it anything else that we wanted, but most programming resources use df for data frames.  I'll stick with that throughout most of the course.  In this case I'm going to include the G and M to distinguish between the two.

- Add a Code cell below this cell using the **+ Code** button.
- In the new Code cell, type these two lines exactly as shown:

  ```python
  dfM = yf.download("MSFT", start="2020-01-01", end="2026-01-01")
  dfG = yf.download("GOOG", start="2020-01-01", end="2026-01-01")
  ```
- Here is what each part of the line does:
  - `yf.download(...)` uses the yfinance library (remember, we shortened it to `yf`) to download stock data.
  - `start="2020-01-01"` is the first date. Dates are written year-month-day.
  - `end="2026-01-01"` is the end date. yfinance stops the day **before** the end date, so to include December 31, 2025 we have to put January 1, 2026.
- Run the cell. You should see a progress bar for each stock that ends with **1 of 1 completed**.
- Now look at the **Variables** pane in the upper right of Positron. You should see **dfM** and **dfG** listed there. Each one is a **data frame**, a table where:
  - Each **row** is one **observation**, which here is one trading day. The date of each day is shown on the left side of the table. Weekends and market holidays are skipped because the stock market is closed.
  - Each **column** is a **variable**, which is a piece of information recorded for every day. The columns include the Open, High, Low and Close prices and the Volume for each day. You'll notice each column name also includes the ticker symbol, for example **Close** and **MSFT**. Prices are adjusted for stock splits, like Google's 20-for-1 split in July 2022.
- Click on **dfM** in the Variables pane to open it and scroll through the data.  It'll open a page on top of this page where you can view all of your data in that data frame. Do the same for **dfG**.  You can close it when you are done looking at it.
- Congratulations! You've downloaded real stock market data into Python and completed Step 4!
- If you need additional help, please [click here](#) for the video guide.


````python
dfM = yf.download("MSFT", start="2020-01-01", end="2026-01-01")
dfG = yf.download("GOOG", start="2020-01-01", end="2026-01-01")
````

## Step 5
Now that we have our data, let's clean it up: keep only the closing prices, drop 2020, join the two data frames, and calculate daily returns.

**Part A: Keep only the closing price and the date**
- Add a Code cell below this cell using the **+ Code** button, then type these two lines:

  ```python
  dfM = dfM["Close"]
  dfG = dfG["Close"]
  ```
- Run the cell. The square brackets `[ ]` select a column from a data frame. By putting `dfM =` in front, we replace the old dfM with a new version that only has the closing price. Notice that we don't have to ask for the date separately. The date is the **index** of the data frame (the labels down the left side), so it always stays with the data.
- Look at dfM in the **Variables** pane. It should now have just one column, and that column is named **MSFT** (the ticker symbol) instead of Close.

**Part B: Drop all 2020 observations**
- Add another Code cell and type:

  ```python
  dfM = dfM.drop(dfM.loc["2020-01-01":"2020-12-31"].index)
  dfG = dfG.drop(dfG.loc["2020-01-01":"2020-12-31"].index)
  ```
- Run the cell. Note the following pieces of those lines of code:
  - `dfM.loc["2020-01-01":"2020-12-31"]` selects the rows by their index label (the date), from January 1, 2020 through December 31, 2020. With `.loc`, both the start and the end dates are included.
  - `.index` takes just the dates of those selected rows.
  - `dfM.drop( ... )` removes the rows with those dates from dfM.
- Check the **Variables** pane again. Note that 2020 is gone.

**Part C: Join the data frames by date**
- Add another Code cell and type:

  ```python
  df = dfM.join(dfG)
  ```
- Run the cell. `.join()` combines two data frames side by side, lining up the rows by their index. Since the index is the date, Microsoft's closing price for a given day ends up on the same row as Google's closing price for that same day. We save the combined data frame as **df**.
- Open **df** in the **Variables** pane. You should see one row for each trading day and two columns, **MSFT** and **GOOG**. Notice that the first date is now January 4, 2021, the first trading day of 2021, because all of 2020 has been removed.
- Before we go on, let's give our columns clearer names. Right now they're just called MSFT and GOOG, but soon we'll add return columns too. That will get confusing if we don't name our variables clearly, so let's add the word "price" to the end of each name. Add another Code cell and type:

  ```python
  df = df.rename(columns={"MSFT": "MSFTprice", "GOOG": "GOOGprice"})
  ```
- Run the cell. `.rename()` changes names in a data frame. `columns=` tells it we're renaming columns, and the curly braces `{ }` pair each old name with its new name: `"oldvar": "newvar"`. If you've had me for an R class in the past (or perhaps will in the future), be careful: the `rename` command in R uses the opposite order, `newvar = oldvar`.
- Check **df** in the **Variables** pane. The columns should now be called **MSFTprice** and **GOOGprice**.

**Part D: Calculate the one-day return**
- A stock's **one-day return** is the percent change in its price from one trading day to the next. 
- Add another Code cell and type:

  ```python
  df["MSFTpc"] = df["MSFTprice"].pct_change()
  df["GOOGpc"] = df["GOOGprice"].pct_change()
  ```
- Run the cell. Here's what the first line does:
  - `df["MSFTprice"]` selects the MSFT closing price column.
  - `.pct_change()` calculates the percent change from each row to the next row, which gives us the one-day return.
  - `df["MSFTpc"] =` creates a new column in df called **MSFTpc** (short for MSFT percent change) and stores the returns there.
- Open **df** in the **Variables** pane. You should now see four columns: **MSFTprice**, **GOOGprice**, **MSFTpc** and **GOOGpc**. A few things to notice:
  - The returns are shown as decimals, not percentages. A value of 0.0259 means a 2.59% gain, and −0.0259 means a 2.59% loss.
  - The first row (January 4, 2021) shows **NaN**, which stands for "Not a Number." This is how pandas marks a missing value. There is no return for the first day because there's no earlier day in our data to use in the calculation.

- Congratulations! You've cleaned and combined your data and completed Step 5!
- If you need additional help, please [click here](#) for the video guide.


````python
dfM = dfM["Close"]
dfG = dfG["Close"]
````

````python
dfM = dfM.drop(dfM.loc["2020-01-01":"2020-12-31"].index)
dfG = dfG.drop(dfG.loc["2020-01-01":"2020-12-31"].index)
````

````python
df = dfM.join(dfG)
````

````python
df = df.rename(columns={"MSFT": "MSFTprice", "GOOG": "GOOGprice"})
````

````python
df["MSFTpc"] = df["MSFTprice"].pct_change()
df["GOOGpc"] = df["GOOGprice"].pct_change()
````

## Step 6
Now let's visualize our data with three graphs: a line graph of prices, a histogram of returns, and a scatter plot comparing the two stocks' returns. We'll use matplotlib, following Chapter 9 of McKinney's *Python for Data Analysis*. Each graph follows the same pattern: create a blank figure, draw on it, then add labels.

**Part A: Line graph of prices**
- Add a Code cell and type:

  ```python
  fig, ax = plt.subplots()
  ax.plot(df["MSFTprice"], label="Microsoft (MSFT)")
  ax.plot(df["GOOGprice"], label="Google (GOOG)")
  ax.legend()
  ax.set_title("Daily Closing Prices, 2021–2025")
  ax.set_xlabel("Date")
  ax.set_ylabel("Price (dollars)")
  fig.subplots_adjust(bottom=0.2)
  fig.text(0.1, 0.02, "Data source: Yahoo Finance. Prices are adjusted for stock splits.", fontsize=8);
  ```
- Run the cell. The graph appears below it. Here's what each line does:
  - `fig, ax = plt.subplots()` creates a blank figure (`fig`, the whole image) with one set of axes (`ax`, the graph area we draw on).
  - `ax.plot(...)` draws a line for one column. The date index becomes the horizontal axis automatically. `label=` names the line for the legend.
  - `ax.legend()` adds the legend box showing which line is which.
  - `ax.set_title`, `ax.set_xlabel` and `ax.set_ylabel` add the title and axis labels.
  - `fig.subplots_adjust(bottom=0.2)` adds space at the bottom of the figure, and `fig.text(...)` writes our footnote there. The two numbers are the position, measured as a fraction of the figure's width and height from the bottom-left corner.
  - The semicolon `;` at the end of the last line hides a line of extra text that Jupyter would otherwise print above the graph.

**Part B: Histogram of Google's one-day returns**
- A **histogram** shows how often values fall into different ranges. Add a Code cell and type:

  ```python
  fig, ax = plt.subplots()
  ax.hist(df["GOOGpc"], bins=50)
  ax.set_title("GOOG Daily Returns, 2021–2025")
  ax.set_xlabel("Percent change in Google (one-day return)")
  ax.set_ylabel("Number of days")
  fig.subplots_adjust(bottom=0.2)
  fig.text(0.1, 0.02, "Data derived from Yahoo Finance price data, adjusted for stock splits.", fontsize=8);
  ```
- Run the cell. `ax.hist(...)` sorts the returns into 50 ranges, called **bins**, and draws a bar showing how many days fall into each one. Notice that most days have returns close to 0, and big moves in either direction are rare. Remember that returns are decimals, so 0.05 on the horizontal axis means 5%.

**Part C: Scatter plot of Google vs. Microsoft returns**
- A **scatter plot** shows the relationship between two variables. Each dot is one trading day. Add a Code cell and type:

  ```python
  fig, ax = plt.subplots()
  ax.scatter(df["GOOGpc"], df["MSFTpc"])
  ax.set_title("One-Day Returns: Google vs. Microsoft, 2021–2025")
  ax.set_xlabel("GOOG")
  ax.set_ylabel("MSFT")
  fig.subplots_adjust(bottom=0.2)
  fig.text(0.1, 0.02, "Data derived from Yahoo Finance price data, adjusted for stock splits.", fontsize=8);
  ```
- Run the cell. In `ax.scatter(...)`, the first column goes on the horizontal axis (Google) and the second goes on the vertical axis (Microsoft). The dots slope upward from lower left to upper right, which means the two stocks tend to move in the same direction on the same day.
